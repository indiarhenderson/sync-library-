```python
tagged_tracks = pd.read_csv('tracks_tagged.csv')
tagged_tracks.head()
```

```python
tagged_tracks.shape
```

```python
tagged_tracks["mood_list"] = tagged_tracks["mood"].fillna("").apply(
    lambda x: [m.strip() for m in x.split(";") if m.strip()]
)
tagged_tracks[["title", "mood", "mood_list"]].head()
```

```python
brief_1 = {"name": "Coming-of-age film",
           "mood": ["warm", "upbeat", "nostalgic", "hopeful", "dreamy"],
           "tempo": ["mid"],
           "min_length": 210, "max_length": 270}      # 3:30 to 4:30

brief_2 = {"name": "Club scene",
           "mood": ["tense", "dark", "gritty", "raw", "euphoric"],
           "tempo": ["fast"],
           "min_length": 60, "max_length": 120}       # 1 to 2 minutes

brief_3 = {"name": "Sad scene",
           "mood": ["emotional", "melancholic", "reflective"],
           "tempo": ["slow"],
           "max_length": 180,                         # under 3 minutes
           "max_instruments": 2}                      # "sparse instrumentation"
```

```python
tagged_tracks["length"] = tagged_tracks["length"].astype(int)
```

```python
def score_track(row, brief):
    score = 0

    # 1 point for each mood word that matches the brief
    for m in row["mood_list"]:
        if m in brief["mood"]:
            score += 1

    # 2 points if the tempo matches
    if row["tempo"] in brief["tempo"]:
        score += 2

    # 1 point if the track has at least the minimum usable length
    min_len = brief.get("min_length")
    if min_len is not None and row["length"] >= min_len:
        score += 1

    # 1 point for sparse instrumentation, if the brief asks for it
    max_inst = brief.get("max_instruments")
    if max_inst is not None:
        n = len([i for i in str(row["instruments"]).split(";") if i and i != "nan"])
        if 0 < n <= max_inst:
            score += 1

    return score
```

```python
def top_matches(brief, n=5):
    tagged_tracks["score"] = tagged_tracks.apply(score_track, axis=1, brief=brief)
    result = tagged_tracks.sort_values("score", ascending=False).head(n)
    return result[["title", "artist", "mood", "tempo", "length", "instruments", "score"]]
```
